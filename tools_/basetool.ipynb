```python
from langchain_core.tools import BaseTool
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
from typing import Type


class Multiply(BaseModel):
    a:int=Field(required=True,description="Enter the 1st value")
    b:int=Field(required=True,description="enter the 2nd value")


class MultiplyTool(BaseTool):
    name:str="multiply"
    description:str="multiply two numbers"
    args_schema:Type[BaseModel]=Multiply
    def _run(self,a:int,b:int):
        return a*b



    

                        
```

```python
multiply_tool=MultiplyTool()
```

```python
result=multiply_tool.invoke({
    'a':4,
    'b':5
})

```

```python
print(result)
print(multiply_tool.name)
print(multiply_tool.args)
```
