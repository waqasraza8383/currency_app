```python
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
import requests
```

```python
@tool

def multiply(a:int,b:int)->int:
    """This tool multiply the a and b"""
    return a*b
```

```python
print(multiply.invoke({'a':10,'b':20}))
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
llm=ChatOpenAI()
```

```python
llm_with_tool=llm.bind([multiply])
```

```python
result=llm_with_tool.invoke("Hello How are you ")
```
