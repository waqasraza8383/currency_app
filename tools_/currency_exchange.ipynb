```python
import requests
import os
from dotenv import load_dotenv
from langchain_core.tools import tool

load_dotenv()

@tool
def compare_currency(
    base_currency: str,
    target_currency: str
) -> str:
    """Compare two currencies and return their exchange rate."""

    api_key = os.getenv("EXCHANGE_RATE_API_KEY")
    url = (
        f"https://v6.exchangerate-api.com/v6/"
        f"{api_key}/latest/{base_currency}"
    )

    response = requests.get(url, timeout=15)
    data = response.json()
    rate = data["conversion_rates"][target_currency]

    return f"1 {base_currency} = {rate} {target_currency}"

# Test the tool
result = compare_currency.invoke({
    "base_currency": "USD",
    "target_currency": "INR"
})

print(result)
```
