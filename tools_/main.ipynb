```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class MultiplyInput(BaseModel):
    a:int=Field(required=True,description="The first number to add")
    b:int=Field(required=True,description='The second number to add')
```

```python
def multiply_func(a:int,b:int):
    return a*b
```

```python
def multiply_func(a:int,b:int)->int:
    return a*b
```

```python
multipy_tool=StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
result=multipy_tool.invoke({'a':3,'b':10})
```

```python
print(result)
print(multipy_tool.name)
print(multipy_tool.description)
```

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
# Step 1: Define input schema
class StudentInput(BaseModel):
    name: str = Field(description="Student name")
    age: int = Field(gt=0, description="Student age")
    marks: float = Field(ge=0, le=100, description="Student marks")

# Step 2: Create Python function
def student_info(name: str, age: int, marks: float) -> str:
    return f"{name} is {age} years old and got {marks} marks."

# Step 3: Convert function into structured tool
student_tool = StructuredTool.from_function(
    func=student_info,
    name="student_information",
    description="Get information about a student.",
    args_schema=StudentInput
)

# Step 4: Invoke tool
result = student_tool.invoke({
    "name": "Ali",
    "age": 20,
    "marks": 85.5
})

print(result)
```
