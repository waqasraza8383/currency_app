```python
class NakiliLLM:
    def __init__(self):
        print("LLM created ")

    def predict(self,prompt):
        responce_list=[
            "India is the capital of pak",
            "Pak is the capital of india",
            "salman is my friend"
        ]
        return {"response":random.choice(responce_list)}
    
```

```python
llm=NakiliLLM()
llm.predict("what is the capital of pakistan")
```

```python
class NakliLLMChain:
    def __init__(self,llm,prompt):
        self.llm=llm
        self.prompt=prompt
    def run(self,input_dict):
        final_prompt=self.prompt.format(input_dict)
        result=self.llm.predict(final_prompt)
        return result['response']

        
```
