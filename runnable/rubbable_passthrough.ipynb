```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.runnables import Runnable,RunnableParallel,RunnableSequence
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import OpenAI,ChatOpenAI,OpenAIEmbeddings
```

```python
passthrough=RunnablePassthrough()
```

```python
prompt1=PromptTemplate(
    template='tell me about the jok{topic}',
    input_variables=['topic']
)
```

```python
model=OpenAI()
```

```python
prompt2=PromptTemplate(
    template='tell me about your school {topic}',
    input_variables=['topic']
)
```

```python
parser=StrOutputParser()
```

```python
joke_gen_chain=RunnableSequence(prompt1,model,parser)
```

```python
parallel=RunnableParallel(
    {
        'joke':RunnablePassthrough(),
        'explanation':RunnableSequence(prompt2,model,parser)
    }
)
```

```python
final_chain=RunnableSequence(parser,parallel)

```

```python
final_chain.invoke({"tell me about the video"})
```
