```python
from langchain_core.runnables import RunnableSequence,RunnableLambda,RunnablePassthrough,RunnableBranch
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import OpenAI,ChatOpenAI,chat_models
```

## creating the prompt

```python
prompt1=PromptTemplate(
    template='Enter your name {topic}',
    input_variables=['topic']
)

```

```python
prompt2=PromptTemplate(
    template='Enter your joke {topic}',
    input_variables=['topic']
)
```

# importing the models

```python
model=OpenAI()
```

```python
parser=StrOutputParser()
```

```python
report_gen_chain=RunnableSequence(prompt1,model,parser)
```

```python
branch_chain=RunnableBranch(
    (lambda x:len(x.split())>500,RunnableSequence(prompt2,model,parser)),
    RunnablePassthrough()

)
```

```python
final_chain=RunnableSequence(report_gen_chain,branch_chain)
```

```python
print(final_chain.invoke({'topic':'RUssia vs ukraine'}))
```
