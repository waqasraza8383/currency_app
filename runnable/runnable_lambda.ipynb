```python
from langchain_core.runnables import RunnableLambda
from langchain_core.runnables import RunnablePassthrough
from langchain_core.runnables import Runnable,RunnableParallel,RunnableSequence
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import OpenAI,ChatOpenAI,OpenAIEmbeddings
```

```python
def word_count(text):
    return  len(text.split)
```

```python
prompt1=PromptTemplate(
    template='tell me a {joke}',
    input_variables=['joke']
)
```

```python
prompt2=PromptTemplate(
    template='tell me about your name {joke}',
    input_variables=['joke']

)
```

```python
model=OpenAI()
```

```python
parser=StrOutputParser()
```

```python

joke_gen=RunnableParallel(prompt1,model,parser)
```

```python
parallel_chain=RunnableParallel({
    'joke':RunnablePassthrough(),
    'word_count':RunnableLambda(word_count)
})
```

```python
def word_count(text):
    return len(text.split())



return_word_counter=RunnableLambda(word_count)


```

```python
return_word_counter.invoke("Hi this is me")
```

```python
final_chain=RunnableSequence(joke_gen,parallel_chain)
```
